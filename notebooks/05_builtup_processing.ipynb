# Built-up indicator
Calculate NDBI only after selecting and documenting suitable SWIR and NIR reflectance bands.

```python
from pathlib import Path
import subprocess, sys
swir, nir = Path('../data/raw/swir.tif'), Path('../data/raw/nir.tif')
if not swir.exists() or not nir.exists():
    raise FileNotFoundError('Add aligned SWIR and NIR bands and document the selected sensor.')
subprocess.run([sys.executable, '../scripts/process_builtup.py', '--swir', str(swir), '--nir', str(nir), '--output', '../gis/raster/chennai_ndbi.tif'], check=True)
```
