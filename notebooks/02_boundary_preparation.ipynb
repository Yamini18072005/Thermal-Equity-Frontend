# Boundary preparation
Validate an exported official GCC vector boundary before permanent storage.

```python
from pathlib import Path
import geopandas as gpd
boundary_path = Path('../gis/boundaries/Region_Boundary.gpkg')
if not boundary_path.exists():
    raise FileNotFoundError('Export the active GCC Region Boundary from QGIS first.')
boundary = gpd.read_file(boundary_path, layer='Region_Boundary')
display(boundary.head())
print(boundary.crs, len(boundary), boundary.total_bounds, (~boundary.geometry.is_valid).sum())
```
