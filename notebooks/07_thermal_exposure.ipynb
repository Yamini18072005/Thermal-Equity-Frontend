# Thermal exposure
Summarize LST by validated ward, zone, or region polygons.

```python
from pathlib import Path
import subprocess, sys
raster = Path('../gis/raster/chennai_lst.tif')
wards = Path('../gis/boundaries/Ward_Boundary.gpkg')
if not raster.exists() or not wards.exists():
    raise FileNotFoundError('Provide chennai_lst.tif and a validated Ward_Boundary.gpkg.')
subprocess.run([sys.executable, '../scripts/thermal_exposure.py', '--raster', str(raster), '--boundaries', str(wards), '--key', 'ward_id', '--output', '../outputs/tables/thermal_exposure_by_ward.csv'], check=True)
```
