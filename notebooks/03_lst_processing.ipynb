# Landsat LST processing
Use a documented Landsat Collection 2 Level-2 ST_B10 product and QA_PIXEL mask.

```python
from pathlib import Path
import subprocess, sys
thermal = Path('../data/raw/landsat_ST_B10.tif')
if not thermal.exists():
    raise FileNotFoundError('Add the selected Landsat ST_B10 product and record it in outputs/data_catalog.csv.')
subprocess.run([sys.executable, '../scripts/process_lst.py', '--thermal', str(thermal), '--output', '../gis/raster/chennai_lst.tif'], check=True)
```
