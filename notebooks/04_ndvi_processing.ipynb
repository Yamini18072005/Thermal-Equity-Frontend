# NDVI processing
Calculate NDVI from aligned cloud-masked red and NIR reflectance bands.

```python
from pathlib import Path
import subprocess, sys
red, nir = Path('../data/raw/red.tif'), Path('../data/raw/nir.tif')
if not red.exists() or not nir.exists():
    raise FileNotFoundError('Add aligned optical bands and document their sensor and date.')
subprocess.run([sys.executable, '../scripts/process_ndvi.py', '--red', str(red), '--nir', str(nir), '--output', '../gis/raster/chennai_ndvi.tif'], check=True)
```
